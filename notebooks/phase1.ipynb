## Read the results

MAE and RMSE measure forecast error in daily-volatility percentage points; lower is better. An error of 0.002 in decimal units equals 0.2 percentage points. Daily forecasts have overlapping future windows. The second score uses forecasts five trading sessions apart, with no overlapping windows. The reserved test period stays unscored.
